```python
from typing import Any, Dict
from dateutil.parser import parse as parse_date, ParserError
from openai import OpenAI
import os, json, time, uuid
from typing import Any, Dict
from openai import OpenAI
from dateutil.parser import parse as parse_date, ParserError
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_community.document_loaders import PyPDFLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
```

```python
# 📂 Paths
# =====================================================
FOLDER_PATH = "/content/wedlii_docs"
INDEX_PATH = "/content/faiss_index"
USERS_FILE = "users.json"
CONVOS_FILE = "conversations.json"
CONVO_BUFFER_SIZE = 10
```

```python
OPENAI_API_KEY = "sk-proj"   # <<< PUT YOUR KEY HERE
client = OpenAI(api_key=OPENAI_API_KEY)
```

```python
embedding = OpenAIEmbeddings(openai_api_key=OPENAI_API_KEY)

if os.path.exists(f"{INDEX_PATH}/index.faiss"):
    print("✅ Loading existing FAISS index...")
    vectorstore = FAISS.load_local(INDEX_PATH, embedding, allow_dangerous_deserialization=True)
else:
    if not os.path.exists(FOLDER_PATH):
        os.makedirs(FOLDER_PATH)
        print(f"📂 Created '{FOLDER_PATH}'. Please add Wedlii PDF files and rerun.")
        raise SystemExit

    all_documents = []
    for file in os.listdir(FOLDER_PATH):
        if not file.lower().endswith(".pdf"):
            print(f"⚠️ Skipping non-PDF file: {file}")
            continue
        try:
            loader = PyPDFLoader(os.path.join(FOLDER_PATH, file))
            docs = loader.load()
            for i, doc in enumerate(docs, 1):
                doc.metadata.update({
                    "source_file": file,
                    "chunk_number": i,
                    "page_number": i
                })
            all_documents.extend(docs)
            print(f"✅ Loaded {len(docs)} pages from {file}")
        except Exception as e:
            print(f"❌ Error loading {file}: {e}")

    if not all_documents:
        print("⚠️ No PDFs found. Exiting.")
        raise SystemExit

    splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=100)
    chunks = splitter.split_documents(all_documents)

    vectorstore = FAISS.from_documents(chunks, embedding)
    vectorstore.save_local(INDEX_PATH)
    print("✅ FAISS index created and saved!")
```

```python
# 📂 Storage Helpers

def load_json(path: str) -> Dict:
    if os.path.exists(path):
        with open(path, "r", encoding="utf-8") as f:
            return json.load(f)
    return {}

def save_json(path: str, data: Dict):
    with open(path, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=2)

users_store = load_json(USERS_FILE)
convos_store = load_json(CONVOS_FILE)
```

```python
# =====================================================
# 👤 User Handling
# =====================================================
def default_user(user_id: str) -> Dict[str, Any]:
    return {
        "id": user_id,
        "first_name": "",
        "last_name": "",
        "wedding_date": "",
        "venue": "",
        "celebrant": "",
        "style": "",
        "guest_count": 0,
        "total_budget": 0.0,
        "memory_summary": ""
    }

def validate_input(field: str, value: str):
    try:
        if field == "wedding_date":
            parse_date(value)
            return value
        if field == "guest_count":
            return int(value)
        if field == "total_budget":
            return float(value)
        return value
    except (ValueError, ParserError):
        return None

def update_user_detail(user_id: str, field: str, value: str):
    valid = validate_input(field, value)
    if valid is not None:
        users_store[user_id][field] = valid
        save_json(USERS_FILE, users_store)
        return True
    return False

def get_user_detail(user_id: str, field: str):
    return users_store[user_id].get(field, "Not set")
```

```python
# =====================================================
# 📝 Conversation Handling
# =====================================================
def append_conversation_turn(user_id: str, user_msg: str, assistant_msg: str):
    history = convos_store.get(user_id, [])
    history.append({
        "user_message": user_msg,
        "assistant_message": assistant_msg,
        "ts": int(time.time())
    })
    convos_store[user_id] = history[-10:]
    save_json(CONVOS_FILE, convos_store)
```

```python
# 🧠 Memory Summary
# =====================================================
def summarize_memory(user_id: str):
    history = convos_store.get(user_id, [])
    if not history:
        return ""
    conversation_text = "\n".join([
        f"User: {turn['user_message']}\nBot: {turn['assistant_message']}"
        for turn in history
    ])
    try:
        resp = client.chat.completions.create(
            model="gpt-4.1-nano",
            messages=[
                {"role": "system", "content": "Summarize this conversation for memory."},
                {"role": "user", "content": conversation_text}
            ],
            max_tokens=200
        )
        summary = resp.choices[0].message.content.strip()
        users_store[user_id]["memory_summary"] = summary
        save_json(USERS_FILE, users_store)
        return summary
    except Exception as e:
        return f"(Memory error: {e})"
```

```python
# 🤖 System Prompt
# =====================================================
def build_system_prompt(user_id: str) -> str:
    user = users_store.get(user_id, default_user(user_id))
    details = "\n".join([f"{k}: {user[k]}" for k in [
        "first_name", "last_name", "wedding_date", "venue",
        "celebrant", "style", "guest_count", "total_budget"
    ]])
    memory = user.get("memory_summary", "")
    return f"""You are Wedlii, a wedding planning assistant.
Here are the user details:
{details}

Conversation memory:
{memory}
"""
```

```python
# 🛠 Functions (tools) using raw dicts
# =====================================================
functions = [
    {
        "name": "update_user_detail",
        "type": "function",
        "description": "Update a wedding detail field for the current user",
        "parameters": {
            "type": "object",
            "properties": {
                "user_id": {"type": "string"},
                "field": {
                    "type": "string",
                    "enum": ["first_name","last_name","wedding_date","venue","celebrant","style","guest_count","total_budget"]
                },
                "value": {"type": "string"}
            },
            "required": ["user_id","field","value"]
        },
    },
    {
        "name": "get_user_detail",
        "type": "function",
        "description": "Retrieve a specific wedding detail for the current user",
        "parameters": {
            "type": "object",
            "properties": {
                "user_id": {"type": "string"},
                "field": {
                    "type": "string",
                    "enum": ["first_name","last_name","wedding_date","venue","celebrant","style","guest_count","total_budget"]
                }
            },
            "required": ["user_id","field"]
        },
    },
]
```

```python

# =====================================================
# 💬 Main Chat Function
# =====================================================
def ask_wedlii(user_id: str, user_text: str, temperature: float = 0.5):
    if user_id not in users_store:
        users_store[user_id] = default_user(user_id)
        save_json(USERS_FILE, users_store)

    system_prompt = build_system_prompt(user_id)

    messages = [
        {"role": "system", "content": system_prompt + "\nTone & Personality: You are warm, friendly, concise."},
        {"role": "user", "content": user_text}
    ]

    try:
        resp = client.chat.completions.create(
            model="gpt-4.1-nano",
            messages=messages,
            functions=functions,
            function_call="auto",
            temperature=temperature
        )

        msg = resp.choices[0].message

        # Handle function calls
        if hasattr(msg, "function_call") and msg.function_call:
            name = msg.function_call.name
            args = json.loads(msg.function_call.arguments)

            # ✅ Force CLI user_id (fix security check)
            args["user_id"] = user_id

            if name == "update_user_detail":
                success = update_user_detail(args["user_id"], args["field"], args["value"])
                reply = f"✅ Updated {args['field']} to {args['value']}." if success else f"⚠️ Could not update {args['field']}."
                append_conversation_turn(user_id, user_text, reply)
                return reply

            if name == "get_user_detail":
                detail = get_user_detail(args["user_id"], args["field"])
                reply = f"📖 Your {args['field']} is: {detail}."
                append_conversation_turn(user_id, user_text, reply)
                return reply

        # Normal assistant reply
        answer = msg.content.strip()
        append_conversation_turn(user_id, user_text, answer)
        summarize_memory(user_id)
        return answer

    except Exception as e:
        return f"❌ Error handling request: {e}"
```

```python
# 🖥 CLI
# =====================================================
if __name__ == "__main__":
    print("Welcome to Wedlii Assistant! Enter user ID or press Enter for new one.")
    uid = input("User ID: ").strip()
    if not uid or uid not in users_store:
        uid = str(uuid.uuid4())
        users_store[uid] = default_user(uid)
        save_json(USERS_FILE, users_store)
        print(f"✨ New user created: {uid}")

    print("\nType 'quit' to exit.\n")
    while True:
        user_in = input("You: ").strip()
        if user_in.lower() in ["quit", "exit"]:
            break
        bot_out = ask_wedlii(uid, user_in)
        print("Wedlii:", bot_out)
```
